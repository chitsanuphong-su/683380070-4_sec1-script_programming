## โค้ดเก่าจาก week2 Lab 2.1

```python
txt = input("Enter your text: ")

num = int(txt)

if num > 0:
  if num % 2 == 0:
    print(f"{num} is Positive and Even.")
  else:
    print(f"{num} is Positive and Odd.")
elif num < 0:
  if num % 2 == 0:
    print(f"{num} is Negative and Even.")
  else:
    print(f"{num} is Negative and Odd.")
else:
  print(f"{num} is Zero.")
```

## Refactoring code

```python
# ฟังก์ชันที่ 1: ตรวจสอบว่าเป็นบวก ลบ หรือศูนย์
def check_sign(num):
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"

# ฟังก์ชันที่ 2: ตรวจสอบว่าเป็นเลขคู่ หรือเลขคี่
def check_even_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

# ฟังก์ชันที่ 3: นำผลลัพธ์จากสองฟังก์ชันบนมาประกอบกัน
def classify_number(num):
    sign = check_sign(num)

    if sign == "Zero":
        return f"{num} is Zero."
    else:
        parity = check_even_odd(num)
        return f"{num} is {sign} and {parity}."


# --- ส่วนการทำงานหลักของโปรแกรม (Main Program) ---
txt = input("Enter your text: ")
num = int(txt)

result = classify_number(num)
print(result)
```
