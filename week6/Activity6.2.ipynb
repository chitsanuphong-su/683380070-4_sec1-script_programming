# Activity 6.2: อภิปรายข้อดีของการเขียนแบบ Module และสำรวจ Standard Library
รูปแบบ: จับกลุ่มคุยกันหรือคุยกันทั้งห้อง
คำสั่ง:

ข้อดีของการแบ่ง Module: ให้อภิปรายกันว่าในสถานการณ์ไหนที่การแตกโค้ดออกเป็นฟังก์ชันย่อยๆ หรือแยกไฟล์ (Module) จะช่วยชีวิตเราได้ (เช่น ตอนทำโปรเจกต์ใหญ่ๆ, ตอนต้องแบ่งงานกับเพื่อน, การหาบั๊ก)

สำรวจ Standard Library: ให้ลองเข้าไปที่เว็บคู่มือ Python (docs.python.org/3/library/) แล้วช่วยกันเลือก Library ที่น่าสนใจมา 2-3 ตัว (เช่น datetime, os, sys, collections) แล้วแชร์ไอเดียว่าเราจะเอามันไปสร้างโปรแกรมอะไรได้บ้าง

### 1. ทำไมเราต้องหั่นโค้ดแยกเป็น Module? (Benefits of Modularity)

สมมติว่าต้องเขียนเกมใหญ่ๆสักเกม หรือทำแอปแบบ Shopee โค้ดโปรเจกต์นึงมีเป็นแสนบรรทัด ถ้ายัดทุกอย่างลงไปในไฟล์ `main.py` ไฟล์เดียว มีมึนหัวจากการรีวิวโค้ดแน่นอน

การที่เราแยกโค้ดเป็นส่วนๆ มันช่วยเราดังนี้:

* **แบ่งงานเพื่อนทำได้ (Teamwork):** สมมติทำโปรเจกต์จบ 3 คน ถ้ามีไฟล์เดียวคือแย่งกันพิมพ์ตีกันตายแน่ๆ แต่ถ้าเราแบ่งเป็น Module เช่น นาย A ไปทำ `login_system.py` นาย B ไปทำ `payment.py` นาย C ทำ `cart.py` แบบนี้ต่างคนต่างทำไฟล์ตัวเองได้เลย แล้วค่อยเอามาประกอบร่างกันทีหลัง
* **หาบั๊กง่าย (Debugging):** สมมติมีคนบอกว่า ระบบตัดบัตรเครดิตพัง เราก็ไปดูแค่ที่โมดูล `payment.py` ได้เลย ไม่ต้องไปนั่งสไลด์เมาส์หาบรรทัดที่เท่าไหร่ก็ไม่รู้ในไฟล์รวมจนตาลาย
* **เอาไปใช้รอบหน้าได้ (Reusability):** สมมติโปรเจกต์นี้เราเขียนโมดูล "ระบบสุ่มกาชา" ออกมาได้ดีมาก พอโปรเจกต์หน้าบอสสั่งทำเกมใหม่ที่มีสุ่มกาชาเหมือนกัน เราก็แค่ก๊อปไฟล์ `gacha_system.py` ไปแปะในโปรเจกต์ใหม่ได้เลย

### 2. Standard Library

Python มีสิ่งที่เรียกว่า Standard Library ซึ่งเป็นเครื่องมือสำเร็จรูปที่ติดมากับภาษาอยู่แล้ว เราไม่ต้องเสียเวลาเขียนโค้ดขึ้นมาใหม่เองทั้งหมดตั้งแต่ศูนย์ แค่ใช้คำสั่ง import ก็นำมาใช้งานได้เลย ช่วยให้เราทำงานไวขึ้นเยอะ

ตัวอย่างโมดูลน่าสนใจที่ดึงมาใช้ได้เลย:

- import datetime (จัดการเรื่องวันเวลา): เอาไว้คำนวณวันและเวลาได้แบบเป๊ะๆ ไม่ต้องมานั่งทดเลขเอง เช่น ทำโปรแกรมคำนวณอายุจากวันเกิด ทำระบบนับถอยหลังวันสอบ หรือบันทึกเวลาที่ลูกค้ากดสั่งซื้อของ

- import os (สั่งงานระบบคอมพิวเตอร์): ช่วยให้โค้ด Python คุยและสั่งการระบบในคอมเราได้ เช่น สั่งสร้างโฟลเดอร์ตามชื่อเดือนอัตโนมัติ ค้นหาไฟล์ที่ต้องการ หรือลบไฟล์ขยะหลายๆ ไฟล์พร้อมกันโดยไม่ต้องมานั่งคลิกเองทีละอัน

- import collections (นับและจัดกลุ่มข้อมูล): มีตัวช่วยเด็ดๆ อย่าง Counter ที่เอาไว้นับข้อมูลซ้ำได้รวดเร็ว เช่น รับรายการอาหารที่เพื่อนเลือกมา แล้วสั่งให้นับสรุปยอดให้ทันทีว่าเมนูไหนกี่จาน โดยไม่ต้องมานั่งเขียนลูปนับเองให้เหนื่อย
