# Lab 6.1: ฟังก์ชันเครื่องคิดเลขพื้นฐาน (Simple Calculator Functions)
วัตถุประสงค์: ให้นักเรียนสร้างและใช้งานฟังก์ชันสำหรับคำนวณทางคณิตศาสตร์เบื้องต้น ฝึกฝนการส่งพารามิเตอร์รับค่า และการส่งค่ากลับ (Return values)

---



```python
def add(a, b):
  ''' ฟังก์ชันสำหรับการบวก '''
  result = a + b
  return result

def subtract(a, b):
  ''' ฟังก์ชันสำหรับการลบ'''
  result = a - b
  return result

def multiply(a, b):
  ''' ฟังก์ชันสำหรับการคูณ '''
  result = a * b
  return result

def divide(a, b):
  ''' ฟังก์ชันสำหรับการหาร '''
  if b == 0:
    print("Error: Division by zero")
    return None

  result = a / b
  return result
```

```python
while True:

  print("=" * 20)
  print("  Simple Calculator")
  print("=" * 20)
  print("1. Add")
  print("2. Subtract")
  print("3. Multiply")
  print("4. Divide")
  print("5. Exit")
  print("=" * 20)

  choice = input("Enter your choice: ")

  if choice == '1':
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    result = add(num1, num2)
    print("Result:", result)
    print()

  elif choice == '2':
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    result = subtract(num1, num2)
    print("Result:", result)
    print()

  elif choice == '3':
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    result = multiply(num1, num2)
    print("Result:", result)
    print()

  elif choice == '4':
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    result = divide(num1, num2)
    print("Result:", result)
    print()

  elif choice == '5':
    print("Exiting the program.\n")
    break

  else:
    print("Invalid choice. Please try again.\n")
```
