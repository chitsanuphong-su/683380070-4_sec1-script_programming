# Lab 5.2: การลงทะเบียนเรียนและการทำงานร่วมกัน (เน้นใช้ Set)
วัตถุประสงค์: จำลองระบบลงทะเบียนเรียน และหานักเรียนที่เรียนซ้ำกันโดยใช้การดำเนินการของ Set (Set Operations)

---



## Initialize Sets

```python
python_students = {"S001", "S003", "S005", "S007", "S009"}
java_students = {"S002", "S003", "S006", "S007", "S010"}
```

## กรณีใช้ attribute ของ set

```python
python_and_java_students = python_students.intersection(java_students)
all_students = python_students.union(java_students)
python_students_only = python_students.difference(java_students)
one_course_students = python_students.symmetric_difference(java_students)

print(f"นักเรียนที่เรียนทั้ง 2 วิชา : {python_and_java_students}")
print(f"นักเรียนทั้งหมด : {all_students}")
print(f"นักเรียนที่เรียนแค่ Python วิชาเดียว : {python_students_only}")
print(f"นักเรียนที่เรียนแค่วิชาเดียว : {one_course_students}")
```

## กรณีใช้ตัวดำเนินการ

```python
python_and_java_students = python_students & java_students
all_students = python_students | java_students
python_students_only = python_students - java_students
one_course_students = python_students ^ java_students

print(f"นักเรียนที่เรียนทั้ง 2 วิชา : {python_and_java_students}")
print(f"นักเรียนทั้งหมด : {all_students}")
print(f"นักเรียนที่เรียนแค่ Python วิชาเดียว : {python_students_only}")
print(f"นักเรียนที่เรียนแค่วิชาเดียว : {one_course_students}")
```

## Remove Duplicates from a List

```python
all_students_list = ["S001", "S003", "S005", "S001", "S002"]

students_sets = set(all_students_list)

all_students_list = list(students_sets)

print(all_students_list)
```
