# Activity 5.1: การประยุกต์ใช้ Dictionary และ Set ในชีวิตจริง
รูปแบบ: กิจกรรมระดมความคิด (กลุ่มย่อย หรือทั้งชั้นเรียน)

วัตถุประสงค์: ทำความเข้าใจการนำ Dictionary และ Set ไปใช้งานจริงในโปรแกรม และฝึกคิดวิเคราะห์ในการเลือกใช้โครงสร้างข้อมูลให้เหมาะสม

---



## 1.Where would you use a dictionary in a game?

สมมติว่าเรากำลังสร้างเกม RPG การใช้ Dictionary จะเหมาะกับตอนที่เราเปิดกระเป๋าเก็บของ (Inventory) ของตัวละครขึ้นมา การใช้โครงสร้างข้อมูลแบบ Dictionary นั้นเหมาะสม เพราะมันจับคู่ "ชื่อของ" กับ "จำนวน" ให้เราเห็นชัดๆ เช่น:
`inventory = {'ยาเติมเลือด': 10, 'ยามะนาว': 5, 'ดาบกากๆ': 1}`
หรือจะเอาไปเก็บสเตตัสตัวละครก็ได้เหมือนกัน เช่น `{'HP': 100, 'MP': 50, 'STR': 20} `พอเราจะใช้ยาเติมเลือด ระบบก็จะวิ่งไปหาคีย์คำว่า 'ยาเติมเลือด' แล้วลดจำนวนลง

---



## 2.How could a dictionary represent data about a product in an online store?

ถ้าเราทำเว็บขายของออนไลน์แบบ Shopee / Lazada การใช้ Dictionary จะเหมาะกับในระบบหลังบ้านของร้านค้า สินค้าเป็นหมื่นๆ แสนๆ ชิ้น เขาจะไม่ใช้ List มาเรียงสินค้า เขาจะใช้ Dictionary โดยเอา "รหัสสินค้า (Product ID)" มาเป็น Key (กุญแจหลัก)
สมมติว่า `{'IT001': {'ชื่อ': 'เมาส์ไร้สาย', 'ราคา': 450, 'สต็อก': 20}}`
เวลาลูกค้ากดตะกร้าสั่งซื้อปุ๊บ ระบบก็แค่หยิบเอารหัส 'IT001' ไปดึงข้อมูลชื่อกับราคาออกมาโชว์ได้เลย

## 3.When would a set be useful in managing user permissions or tags?

ระบบแฮชแท็ก (Hashtag) หรือระบบกำหนดสิทธิ์การใช้งาน การใช้ Set จะเหมาะสมที่สุด จุดเด่นของมันคือ "ห้ามมีของซ้ำ"
สมมติเราทำแอปโซเชียล แล้วยูสเซอร์พิมพ์แคปชั่นว่า *"กินข้าวกัน #อร่อย #หิว #อร่อย"* สังเกตว่ามีแท็ก #อร่อย ซ้ำกัน ถ้าเราโยนพวกนี้เข้า Set ระบบจะยุบ #อร่อย เหลือแค่ตัวเดียวทันที ช่วยประหยัดพื้นที่เซิร์ฟเวอร์ได้เยอะ
หรืออย่างระบบแอดมินเพจ สิทธิ์ในการเข้าถึง (เช่น สิทธิ์โพสต์, สิทธิ์ลบข้อความ) มันไม่ควรมีสิทธิ์ซ้ำซ้อนกัน การใช้ Set เก็บสิทธิ์พวกนี้จะทำให้จัดการง่ายที่สุด

## 4.Can you think of a scenario where you'd need to quickly check if an item exists in a large collection without caring about its order?

ระบบแบนไอพี (Blocked IP) หรือเช็คคนเข้าเว็บ การใช้ Set จะเหมาะที่สุด ถ้าเราสร้างเว็บมาเว็บนึง แล้วมีคนพยายามแฮกเข้ามา เราเลยบล็อก IP พวกนั้นไว้ สมมติว่ามี IP โดนบล็อกอยู่ 100,000 หมายเลข!
เวลาที่มีคนพยายามล็อกอิน เราแค่ต้องการเช็คว่า IP นี้นิสัยไม่ดี หรือเปล่า? เราไม่ได้สนใจเลยว่า IP นี้โดนบล็อกเป็นลำดับที่เท่าไหร่ (ลำดับก่อน-หลังไม่มีผล) เราสนแค่ มันอยู่ในบัญชีดำไหม?
การใช้ Set เช็คเงื่อนไข (เช่น `if ip in blocked_ips:`) มันจะทำงานได้เร็วกว่าการใช้ List มาก เพราะระบบของ Set ถูกออกแบบมาให้ค้นหาข้อมูลประเภทนี้โดยเฉพาะ
