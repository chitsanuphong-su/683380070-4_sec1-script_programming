# Lab 5.1: ระบบจัดการสมุดติดต่ออย่างง่าย (เน้นใช้ Dictionary)
วัตถุประสงค์: สร้างโปรแกรมสมุดติดต่อแบบข้อความที่สามารถ เพิ่ม, ดู, อัปเดต และลบข้อมูลติดต่อได้โดยใช้ Dictionary

---



```python
contacts = {}


def show_menu():
    print("=========== Menu ===========")
    print("1. Add/Update Contact")
    print("2. View Contacts Details")
    print("3. Show All Contacts")
    print("4. Delete Contacts")
    print("5. Exit")
    print("============================")


while True:
    show_menu()
    choice = input("\nChoose your choice: ")

    if choice == "1":
        name = input("Enter contact name: ").strip().title()
        phone = input("Enter phone number: ").strip()
        email = input("Enter email: ").strip()

        contacts[name] = {"phone": phone, "email": email}
        print(f"Successfully saved contact for {name}!\n")

    elif choice == "2":
        view_name = input("Enter contact name to view: ").strip().title()

        details = contacts.get(view_name)

        if details:
            print("\n- Contact found! -")
            print("Name :", view_name)
            print("Phone:", details["phone"])
            print("Email:", details["email"])
        else:
            print("Contact not found!")
        print()

    elif choice == "3":
        if not contacts:
            print("\nNo contacts in your book yet.")
        else:
            print("\n--- All Contacts ---")
            for name, details in contacts.items():
                print("Name :", name)
                print("Phone:", details["phone"])
                print("Email:", details["email"])
                print("-" * 20)
        print()

    elif choice == "4":
        remove_name = input("Enter contact name to delete: ").strip().title()

        if remove_name in contacts:
            contacts.pop(remove_name)
            print(f"Contact '{remove_name}' removed!")
        else:
            print("Contact not found!")
        print()

    elif choice == "5":
        print("Exit program")
        break

    else:
        print("Invalid Choice Please Choose 1-5\n")
```
