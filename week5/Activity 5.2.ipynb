# Activity 5.2: โปรแกรมนับความถี่ของคำ (Word Frequency Counter)
รูปแบบ: โจทย์ท้าทายการเขียนโค้ดเดี่ยว

วัตถุประสงค์: นำการทำงานของ Dictionary ไปประยุกต์ใช้กับการจัดการข้อความ เพื่อนับจำนวนคำและดึงข้อมูลได้อย่างมีประสิทธิภาพ

---



```python
sentence = input("Enter a sentence: ")
sentence = sentence.lower()

words = sentence.split()

words_count = {}

for word in words:
  if word in words_count:
    words_count[word] += 1
  else:
    words_count[word] = 1

print(words_count)
```

```python
sentence = input("Enter a sentence: ")
sentence = sentence.lower()

words = sentence.split()

words_count = {}

for word in words:
  clean_word = word.strip('.,!?"')

  if clean_word:
    if clean_word in words_count:
      words_count[clean_word] += 1
    else:
      words_count[clean_word] = 1

print(words_count)
```
